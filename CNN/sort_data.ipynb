```python
import os 
import numpy as np
from tqdm import tqdm
```

```python

import re

folder_path = "./data"
np.random.seed(188)


MDD_train, MDD_val, MDD_test = [], [], []
H_train, H_val, H_test = [], [], []


MDD_subjects = np.arange(1, 35)
H_subjects   = np.arange(1, 31)


train_pct, val_pct, test_pct = 0.75, 0.125, 0.125


MDD_n_test = int(len(MDD_subjects) * test_pct)
MDD_n_val = int(len(MDD_subjects) * val_pct)

H_n_test = int(len(H_subjects) * test_pct)
H_n_val = int(len(H_subjects) * val_pct)


MDD_sub_test = np.random.choice(MDD_subjects, MDD_n_test, replace=False)
MDD_remaining = np.setdiff1d(MDD_subjects, MDD_sub_test)
MDD_sub_val = np.random.choice(MDD_remaining, MDD_n_val, replace=False)
MDD_sub_train = np.setdiff1d(MDD_remaining, MDD_sub_val)

H_sub_test = np.random.choice(H_subjects, H_n_test, replace=False)
H_remaining = np.setdiff1d(H_subjects, H_sub_test)
H_sub_val = np.random.choice(H_remaining, H_n_val, replace=False)
H_sub_train = np.setdiff1d(H_remaining, H_sub_val)

print("MDD subjects:")
print(f"  Train: {MDD_sub_train}")
print(f"  Val:   {MDD_sub_val}")
print(f"  Test:  {MDD_sub_test}\n")

print("Healthy subjects:")
print(f"  Train: {H_sub_train}")
print(f"  Val:   {H_sub_val}")
print(f"  Test:  {H_sub_test}\n")

def subject_in_filename(filename, subjects):
    for sub in subjects:

        if re.search(rf"\bS{sub}\b", filename):
            return True
    return False


files = [f for f in os.listdir(folder_path) if f.endswith(".npy")]
for filename in tqdm(files, desc="Loading files"):
    data = np.load(os.path.join(folder_path, filename))

    # MDD group
    if filename.startswith("MDD"):
        if subject_in_filename(filename, MDD_sub_train):
            MDD_train.append(data)
        if subject_in_filename(filename, MDD_sub_val):
            MDD_val.append(data)
        if subject_in_filename(filename, MDD_sub_test):
            MDD_test.append(data)

    if filename.startswith("H"):
        if subject_in_filename(filename, H_sub_train):
            H_train.append(data)
        if subject_in_filename(filename, H_sub_val):
            H_val.append(data)
        if subject_in_filename(filename, H_sub_test):
            H_test.append(data)

#copilot used here
print(f"Number of training samples in MDD group: {len(MDD_train)}")
print(f"Number of validation samples in MDD group: {len(MDD_val)}")
print(f"Number of test samples in MDD group: {len(MDD_test)}")
print(f"Number of training samples in Healthy group: {len(H_train)}")
print(f"Number of validation samples in Healthy group: {len(H_val)}")
print(f"Number of test samples in Healthy group: {len(H_test)}")
#copilot used here

np.save("./data/processed_data/H_train_data.npy", H_train)
np.save("./data/processed_data/H_val_data.npy", H_val)
np.save("./data/processed_data/H_test_data.npy", H_test)

np.save("./data/processed_data/MDD_train_data.npy", MDD_train)
np.save("./data/processed_data/MDD_val_data.npy", MDD_val)
np.save("./data/processed_data/MDD_test_data.npy", MDD_test)

```
